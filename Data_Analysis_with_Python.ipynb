```python
import pandas as pd

df = pd.read_excel("ELNIDO_CORON_JANUARY 2026.xlsx", skiprows=5)

print(df.columns)
df.head()
```

```python
df[['Age','Gender']] = df['Age/Gen'].str.split('/', expand=True)

df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

df.head()
```

```python
df['Gender'] = df['Gender'].map({
    'M': 'Male',
    'F': 'Female'
})
```

```python
df = df.drop(columns=['Age/Gen'])
```

```python
df.to_excel("cleaned_passengers.xlsx", index=False)
```
